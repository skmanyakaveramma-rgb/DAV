```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv('/content/StudentPerformanceFactors.csv')

print("Dataset loaded successfully!")
print(df.head())
```

```python

print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python

df.describe()
```

```python

print("Missing Values:")
print(df.isnull().sum())
```

```python

print("Duplicate Rows:", df.duplicated().sum())
```

```python

# Fill missing values in categorical columns
df['Teacher_Quality'].fillna(df['Teacher_Quality'].mode()[0], inplace=True)

df['Parental_Education_Level'].fillna(df['Parental_Education_Level'].mode()[0], inplace=True)

# Fill missing values in Distance_from_Home
df['Distance_from_Home'].fillna(df['Distance_from_Home'].mode()[0], inplace=True)

# Check missing values again
print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python

plt.figure(figsize=(8,5))

plt.scatter(df['Hours_Studied'], df['Exam_Score'])

plt.title('Study Hours vs Exam Score')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')

plt.show()
```

```python

plt.figure(figsize=(8,5))

plt.scatter(df['Attendance'], df['Exam_Score'])

plt.title('Attendance vs Exam Score')
plt.xlabel('Attendance (%)')
plt.ylabel('Exam Score')

plt.show()
```

```python

plt.figure(figsize=(8,5))

sns.histplot(df['Exam_Score'], bins=20, kde=True)

plt.title('Distribution of Exam Scores')
plt.xlabel('Exam Score')
plt.ylabel('Number of Students')

plt.show()
```

```python

plt.figure(figsize=(8,5))

sns.barplot(x='Motivation_Level',
            y='Exam_Score',
            data=df)

plt.title('Motivation Level vs Exam Score')
plt.show()
```

```python

print("Average Exam Score:")
print(df['Exam_Score'].mean())

print("\nHighest Exam Score:")
print(df['Exam_Score'].max())

print("\nLowest Exam Score:")
print(df['Exam_Score'].min())
```

```python

print("Average Exam Score based on Study Hours:")

print(df.groupby(
    pd.cut(df['Hours_Studied'], bins=4)
)['Exam_Score'].mean())
```

```python

correlation = df[['Hours_Studied',
                  'Attendance',
                  'Sleep_Hours',
                  'Previous_Scores',
                  'Exam_Score']].corr()

print(correlation)

plt.figure(figsize=(8,5))
sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title("Correlation Between Student Factors")
plt.show()
```
