```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"/content/StudentsPerformance.csv")
```

```python
print("First 5 rows of the dataset:")
print(df.head())
print("\nDataset shape:")
print(df.shape)
```

```python
data = df["math score"].dropna()
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
print("\nIQR METHOD")
print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
print("Lower Limit:",lower_limit)
print("Upper Limit:",upper_limit)
```

```python
iqr_outliers = df[
    (df["math score"] < lower_limit) |
     (df["math score"] > upper_limit)
]
print("Number of IQR outliers:", len(iqr_outliers))
```

```python
mean = data.mean()
std = data.std()
df["z_score"] = (df["math score"] - mean) / std
z_outliers = df[abs(df["z_score"]) > 3]
print("\nZ-SCORE METHOD")
print("Mean:",mean)
print("Standard Deviation:",std)
```

```python
df["math_score_treated"] = df["math score"].clip(
    lower = lower_limit,
    upper = upper_limit
)
print("Treatmeant method: Capping using IQR limits")
```

```python
comparison = df[
    ["math score", "math_score_treated"]
]
print(comparison.head(20))
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.boxplot(df["math score"].dropna())
plt.title("math score - After Treatment")
plt.ylabel("math score")
plt.subplot(1, 2, 2)
plt.boxplot(df["math_score_treated"].dropna())
plt.title("math score - After Treatment")
plt.ylabel("math score")
plt.show()
```
